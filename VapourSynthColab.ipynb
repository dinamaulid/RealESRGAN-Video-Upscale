```
#@title Setup {display-mode: "form"}
#@markdown Run this to install VapourSynth, VapourSynth plugins and scripts, as well as some example upscaling models.
#@markdown Fixed for Python 3.12.11, CUDA 12.4, and Colab 2025 updates.

# VapourSynth full installer: libzimg, VapourSynth R72, TDeintMod, TTempSmooth
# Safe for Colab / Ubuntu 22.04 (patched to avoid setuptools restart loop)

import os, sys, shutil, tempfile, time
from threading import Thread
from datetime import datetime
from IPython import display

# Ensure we're at root for predictable paths (Colab-friendly)
try:
    os.chdir("/")
except Exception:
    pass

# ---------------------- Helpers ----------------------
def copytree(src, dst, symlinks=False, ignore=None):
    if not os.path.exists(src):
        print(f"Error: Source directory {src} does not exist")
        return
    os.makedirs(dst, exist_ok=True)
    for item in os.listdir(src):
        s = os.path.join(src, item)
        d = os.path.join(dst, item)
        if os.path.isdir(s):
            shutil.copytree(s, d, symlinks, ignore)
        else:
            shutil.copy2(s, d)

def makesrcd(name):
    os.makedirs(f"/src/{name}", exist_ok=True)
    os.chdir(f"/src/{name}")

def mesongit(giturl):
    p = os.path.basename(giturl)[:-4]
    makesrcd(p)
    try:
        if not os.path.exists(p):
            os.system(f"git clone {giturl} {p} || true")
        os.chdir(p)
        if os.path.exists("build"):
            shutil.rmtree("build", ignore_errors=True)
        os.system("meson build")
        os.system("ninja -C build")
        os.system("ninja -C build install")
    except Exception as e:
        print(f"Failed to build {p}: {e}")
    finally:
        os.chdir("/")

# ---------------------- Stages ----------------------
def getdep1():
    # Wait for dpkg lock, update, and install core packages.
    os.chdir("/")
    os.system("while fuser /var/lib/dpkg/lock-frontend >/dev/null 2>&1; do echo 'Waiting for dpkg lock...'; sleep 1; done")
    os.system("apt-get update -y")
    os.system("apt-get install -y --no-install-recommends --fix-broken \\\"\\\n    software-properties-common ninja-build meson build-essential \\\"\\\n    python3-dev libc6-dev autoconf automake libtool pkg-config \\\"\\\n    ffmpeg yt-dlp zlib1g-dev libpython3-dev libtool-bin \\\"\\\n    cmake cmake-data python3-pip python3-setuptools python3-wheel \\\"\\\n    libfftw3-dev libfftw3-single3 wget git p7zip-full")

    # Do NOT downgrade setuptools in Colab; upgrade wheel only for stability
    os.system("pip install -U wheel")

def download_fatpack_scripts():
    os.chdir("/")
    print("Downloading VS FatPack Scripts (optional)...")
    dlurl = "https://github.com/theChaosCoder/vapoursynth-portable-FATPACK/releases/download/r3/VapourSynth64Portable_2019_11_02.7z"
    with tempfile.TemporaryDirectory() as t:
        dpath = os.path.join(t, "VapourSynth64Portable.7z")
        try:
            os.system(f"wget -q --show-progress {dlurl} -O {dpath}")
            os.system(f"7z x -o{t} {dpath} || true")
            scriptsd = os.path.abspath(os.path.join(t, "VapourSynth64Portable", "Scripts"))
            s = os.path.normpath("VapourSynthImports")
            os.makedirs(s, exist_ok=True)
            if os.path.exists(scriptsd):
                copytree(scriptsd, s)
                sys.path.append(s)
        except Exception as e:
            print("Failed to download/extract fatpack:", e)
    # optional community scripts (best-effort)
    try:
        os.system("wget -q -O /VapourSynthImports/muvsfunc_numpy.py https://raw.githubusercontent.com/WolframRhodium/muvsfunc/master/Collections/muvsfunc_numpy.py || true")
    except Exception:
        pass

def getzimg():
    os.chdir("/")
    os.system("rm -rf /tmp/zimg")
    os.system("git clone https://github.com/sekrit-twc/zimg.git /tmp/zimg || true")
    os.chdir("/tmp/zimg")
    os.system("git checkout release-3.0.5 || true")
    os.system("sed -i 's/LT_INIT(win32-dll)/LT_INIT/' configure.ac || true")
    os.system("apt-get install -y autoconf automake libtool || true")
    os.system("autoreconf -fi 2>&1 | tee /tmp/zimg_autogen.log || true")
    os.system("./configure --prefix=/usr/local 2>&1 | tee /tmp/zimg_configure.log || true")
    os.system("make -j$(nproc) 2>&1 | tee /tmp/zimg_make.log || true")
    os.system("make install 2>&1 | tee /tmp/zimg_make_install.log || true")
    os.system("ldconfig /usr/local/lib || true")
    os.chdir("/")

def getvs():
    os.chdir("/")
    os.system("rm -rf /tmp/vapoursynth || true")
    os.system("git clone https://github.com/vapoursynth/vapoursynth.git /tmp/vapoursynth || true")
    os.chdir("/tmp/vapoursynth")
    os.system("git checkout R72 || true")
    os.system("./autogen.sh 2>&1 | tee /tmp/vapoursynth_autogen.log || true")
    # Ensure pkg-config sees /usr/local
    os.environ["PKG_CONFIG_PATH"] = "/usr/local/lib/pkgconfig:/usr/lib/x86_64-linux-gnu/pkgconfig:/usr/lib/pkgconfig:" + os.environ.get("PKG_CONFIG_PATH", "")
    os.system("./configure --prefix=/usr/local --enable-python-module CFLAGS='-I/usr/local/include' LDFLAGS='-L/usr/local/lib' 2>&1 | tee /tmp/vapoursynth_configure.log || true")
    os.system("make -j$(nproc) 2>&1 | tee /tmp/vapoursynth_make.log || true")
    os.system("make install 2>&1 | tee /tmp/vapoursynth_make_install.log || true")
    os.system("ldconfig /usr/local/lib || true")

    # Ensure Python can find compiled module. Add to sys.path and PYTHONPATH.
    sp = "/usr/local/lib/python3.12/site-packages"
    if sp not in sys.path:
        sys.path.append(sp)
    os.environ["PYTHONPATH"] = sp + ":" + os.environ.get("PYTHONPATH", "")

    os.chdir("/")

def getvsplugins():
    os.chdir("/")
    os.system("apt-get install -y meson ninja-build || true")

    # TDeintMod
    try:
        os.system("rm -rf /tmp/TDeintMod || true")
        os.system("git clone https://github.com/HomeOfVapourSynthEvolution/VapourSynth-TDeintMod.git /tmp/TDeintMod || true")
        os.chdir("/tmp/TDeintMod")
        os.system("meson build || true")
        os.system("ninja -C build || true")
        os.system("ninja -C build install || true")
    except Exception as e:
        print("Failed to build TDeintMod:", e)
    os.chdir("/")

    # TTempSmooth (ensure fftw single precision)
    try:
        os.system("rm -rf /tmp/TTempSmooth || true")
        os.system("git clone https://github.com/HomeOfVapourSynthEvolution/VapourSynth-TTempSmooth.git /tmp/TTempSmooth || true")
        os.chdir("/tmp/TTempSmooth")
        os.system("apt-get install -y libfftw3-dev libfftw3-single3 || true")
        os.system("meson build || true")
        os.system("ninja -C build || true")
        os.system("ninja -C build install || true")
    except Exception as e:
        print("Failed to build TTempSmooth:", e)
    os.chdir("/")

    os.makedirs("/usr/local/lib/vapoursynth", exist_ok=True)
    os.system("ldconfig /usr/local/lib/vapoursynth || true")

def getpythonstuff():
    os.chdir("/")
    os.environ["PKG_CONFIG_PATH"] = "/usr/local/lib/pkgconfig:/usr/lib/x86_64-linux-gnu/pkgconfig:/usr/lib/pkgconfig:" + os.environ.get("PKG_CONFIG_PATH", "")
    os.environ["LD_LIBRARY_PATH"] = "/usr/local/lib:/usr/lib/x86_64-linux-gnu:/usr/lib:" + os.environ.get("LD_LIBRARY_PATH", "")
    os.system("pip install -U wheel || true")
    os.system("pip install cython opencv-python || true")
    # IMPORTANT: Do NOT pip-install vapoursynth from PyPI; we built from source above.

def cudastuff():
    os.chdir("/")
    try:
        os.system("pip install git+https://github.com/AlphaAtlas/VSGAN.git || true")
        os.makedirs("/NeuralNetworks", exist_ok=True)
    except Exception as e:
        print("Optional CUDA install failed:", e)

# Propagating thread wrapper
class PropagatingThread(Thread):
    def run(self):
        self.exc = None
        try:
            if hasattr(self, '_Thread__target'):
                self.ret = self._Thread__target(*self._Thread__args, **self._Thread__kwargs)
            else:
                self.ret = self._target(*self._args, **self._kwargs)
        except BaseException as e:
            self.exc = e
    def join(self):
        super(PropagatingThread, self).join()
        if self.exc:
            raise self.exc
        return self.ret

# ---------------------- Run sequence ----------------------
t1 = PropagatingThread(target=getdep1)
t1.start(); print("apt init thread started")

t2 = PropagatingThread(target=download_fatpack_scripts)
t2.start(); print("VS script downloader thread started.")

t3 = PropagatingThread(target=getzimg)
t3.start(); print("zimg downloader thread started.")

# Detect GPU
accelerator = None
gpu = None
if 'COLAB_TPU_ADDR' in os.environ:
    raise Exception("TPUs are (currently) not supported! Please use a GPU or CPU instance.")
else:
    out = os.popen("command -v nvidia-smi").read().strip()
    if out != "":
        try:
            raw = os.popen("nvidia-smi --query-gpu=gpu_name --format=csv").read().splitlines()
            if len(raw) > 1:
                accelerator = "CUDA"
                gpu = raw[1]
                print("Nvidia GPU detected:", gpu)
        except Exception:
            pass
if accelerator == None:
    print("Warning: No Accelerator Detected!")

t1.join(); print("Apt init thread done.")
t3.join(); print("zimg thread done.")

t1 = PropagatingThread(target=getvs)
t1.start(); print("Vapoursynth downloader thread started.")
t1.join(); print("Vapoursynth built & installed")

t3 = PropagatingThread(target=getpythonstuff)
t3.start(); print("Pip thread started")

t1 = PropagatingThread(target=getvsplugins)
t1.start(); print("VS plugin downloader thread started.")

if accelerator == "CUDA":
    t4 = PropagatingThread(target=cudastuff)
    t4.start(); print("CUDA pip thread started.")

t2.join(); print("VS script downloader thread done.")
t3.join(); print("pip thread done")
if accelerator == "CUDA":
    t4.join(); print("CUDA pip thread done.")
t1.join(); print("VS plugin thread done.")

# Additional meson-based plugins
mesongit("https://github.com/HomeOfVapourSynthEvolution/VapourSynth-DCTFilter.git")
mesongit("https://github.com/HomeOfVapourSynthEvolution/VapourSynth-TTempSmooth.git")

# ---------------------- Post-install test ----------------------
sp = "/usr/local/lib/python3.12/site-packages"
if sp not in sys.path:
    sys.path.append(sp)
os.environ["PYTHONPATH"] = sp + ":" + os.environ.get("PYTHONPATH", "")

   # --- Quick import/test ---
try:
    import vapoursynth as vs
    core = vs.core
    print("\n--- Quick import/test ---")
    print("VapourSynth import OK, version:", vs.__version__)

    # Tambahan sederhana
    print("VapourSynth OK:", vs.__version__)

except Exception as e:
    print("Import or test failed:", e)

print("\n✅ Installer script finished (see above for warnings/errors).")
```

```
%%writefile /content/autogenerated.vpy

#This is the Vapoursynth Script!
#Running this cell will write the code in this cell to disk, for VSPipe to read.
#Later cells will check to see if it executes.
#Edit it just like a regular python VS script.
#Search for functions and function reference in http://vsdb.top/, or browse the "VapourSynthImports" folder.

#Import functions
import sys, os, cv2
sys.path.append('/VapourSynthImports')
import vapoursynth as vs
import vsgan as VSGAN
import mvsfunc as mvf
#import muvsfunc as muf
#import fvsfunc as fvf
import havsfunc as haf
import Alpha_CuPy as ape
import muvsfunc_numpy as mufnp
#import BMToolkit as bm
import G41Fun as G41
#import vsutil as util
#import edi_rpow2 as edi
#import kagefunc as kage
#import lostfunc as lost
#import vsTAAmbk as taa
#import xvs as xvs
from vapoursynth import core

#Set RAM cache size, in MB
core.max_cache_size = 10500

#Get Video(s) or Image(s). ffms2 (ffmpeg) or imwri (imagemagick) will read just about anything.
#Lsmash sometimes works if ffms2 failes, d2v reads mpeg2 files
clip = core.ffms2.Source(r"/content/enhance.webm")
#clip = core.lsmas.LWLibavSource("/tmp/%d.png")
#clip = core.imwri.Read("testimage.tiff")

#Store source for previewing
src = clip

#Convert to 16 bit YUV for preprocessing
#clip = core.resize.Spline36(clip, format = vs.YUV444P16)

#Deinterlace
#clip = G41.QTGMC(clip, Preset='Medium')

#Mild deblocking
#clip = fvf.AutoDeblock(clip)

#Convert to floating point RGB
clip = mvf.ToRGB(clip, depth = 32)

#Spatio-temportal GPU denoiser. https://github.com/Khanattila/KNLMeansCL/wiki/Filter-description
clip = core.knlm.KNLMeansCL(clip, a = 8, d = 4, h = 1.4)


preupscale = clip
#Run ESRGAN model. See https://upscale.wiki/wiki/Model_Database
vsgan_device = VSGAN.VSGAN()
vsgan_device.load_model(model=r"/content/4X_Box.pth", scale=4)
clip = vsgan_device.run(clip=clip, chunk = False,  pad = 16)

clip = core.knlm.KNLMeansCL(clip, a = 7, d = 3, h = 1.4)



#Run MXNet model. See the "MXNet" cell.
#Tensorflow models are also supported!
#sr_args = dict(model_filename=r'/NeuralNetworks/ARAN/aran_c0_s1_x4', up_scale=4, device_id=0, block_w=256, block_h=128, is_rgb_model=True, pad=None, crop=None, pre_upscale=False)
#clip = mufnp.super_resolution(clip, **sr_args)

#HQ downscale on the GPU with dpid
#clip = ape.GPU_Downscale(clip, width = 3840, height = 2160)

#Convert back to YUV 444 format/Rec 709 colorspace
clip = core.resize.Spline36(clip, format = vs.YUV444P16, matrix_s = "709")

#Strong temporal denoiser and stabilizer with the LR as a motion reference clip, for stabilizing.
prefilter = core.resize.Spline36(preupscale, format = clip.format, width = clip.width, height = clip.height, matrix_s = "709")
clip = G41.SMDegrain(clip, tr=3, RefineMotion=True, pel = 1, prefilter = prefilter)

#Another CPU denoiser/stabilizer. "very high" is very slow.
#clip = haf.MCTemporalDenoise(clip, settings = "very high", useTTmpSm = True, maxr=4, stabilize = True)

#Stabilized Anti Aliasing, with some GPU acceleration
#clip = taa.TAAmbk(clip, opencl=True, stabilize = 3)

#Example sharpeners that work well on high-res images
#Masks or mvf.limitfilter are good ways to keep artifacts in check
#clip = core.warp.AWarpSharp2(clip)
#clip = G41.NonlinUSM(clip, z=3, sstr=0.28, rad=9, power=1)

#High quality, strong debanding
#clip = fvf.GradFun3(clip, smode = 2)

#Convert back to 8 bit YUV420 for output.
clip = core.resize.Spline36(clip, format = vs.YUV420P8, matrix_s = "709", dither_type = "error_diffusion")

#Interpolate to double the source framerate
#super = core.mv.Super(inter)
#backward_vectors = core.mv.Analyse(super, isb = True,  overlap=4, search = 3)
#forward_vectors = core.mv.Analyse(super, isb = False, overlap=4, search = 3)
#inter = core.mv.FlowFPS(inter, super, backward_vectors, forward_vectors, num=0, den=0)

#Stack the source on top of the processed clip for comparison
src = core.resize.Point(src, width = clip.width, height = clip.height, format = clip.format)
#clip = core.std.StackVertical([clip, src])
#Alternatively, interleave the source and slow down the framerate for easy comparison.
clip = core.std.Interleave([clip, src])
clip = core.std.AssumeFPS(clip, fpsnum = 2)

#clip = core.std.SelectEvery(clip=clip, cycle=48, offsets=[0,1])

clip.set_output()
```
